```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()
```

```python
llm = ChatGroq(
    model = 'openai/gpt-oss-120b',
    max_tokens=512,
    temperature=0
)
```

```python
model = llm
```

```python
template1 = PromptTemplate(
    template='write a detailed report on {topic}',
    input_variables=['topic']
)
```

```python
template2 = PromptTemplate(
    template = 'write a 5 line summary on the fllowing text. /n {text}',
    input_variable=['text']
)
```

```python
parser = StrOutputParser() 
```

```python
chain = template1 | model | parser | template2 | model | parser # to create pipeline to execute flow of steps
# user sends sonething...prompt is formed and sent to model...model processes and gives result with various fields..parser gets the text(detailed report) from result with is then sent to template 2...and again sent to model to generate 5 line summary and meta data...as we dont need meta data we again call the parser
```

```python
result = chain.invoke({'topic':'black hole'})
```
